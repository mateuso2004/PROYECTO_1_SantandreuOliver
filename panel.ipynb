```python
import json

with open('data.json', 'r') as file:
    data = json.load(file)
```

```python
# Producto más ingresos
for listaventas, ventas in data.items():
    ingreso_por_producto = {}
    for venta in ventas:
        producto = venta["producto"]
        ingreso_producto = venta["cantidad"] * venta["precio_unitario"]
        if producto not in ingreso_por_producto:
            ingreso_por_producto[producto] = 0
        ingreso_por_producto[producto] += ingreso_producto

print(ingreso_por_producto)

ingreso_max_producto = 0
productos_max_ingreso = [] #lista por posibles empates
for producto, ingreso in ingreso_por_producto.items():
    if ingreso > ingreso_max_producto:
        ingreso_max_producto = ingreso
        productos_max_ingreso = [producto]
    elif ingreso == ingreso_max_producto: #caso empate
        productos_max_ingreso.append(producto)

if len(productos_max_ingreso) == 1:
    print(f"El producto que más ingresos ha generado ha sido {productos_max_ingreso[0]} con {ingreso_max_producto} €.")
else:
    print(f"Los {len(productos_max_ingreso)} productos que más ingresos han generado (empate) han sido {productos_max_ingreso} con {ingreso_max_producto} € cada uno.")
```

```python
# Filtrar ventas por un vendedor concreto
vendedor_objetivo = "Ana"
ventas_vendedor = []

for listaventas, ventas in data.items():
    for venta in ventas:
        if venta["vendedor"] == vendedor_objetivo:
            ventas_vendedor.append(venta)

print(f"Ventas de {vendedor_objetivo} ({len(ventas_vendedor)}):")
for venta in ventas_vendedor:
    print(venta)
```

```python
# Calcular el ticket medio

for listaventas, ventas in data.items():
    total_facturado = 0.0
    for venta in ventas:
        total_facturado = total_facturado + venta["precio_unitario"] * venta["cantidad"]
    ticket_medio = total_facturado / len(ventas)
    print(f"El ticket medio ha sido {ticket_medio} €.")
```
